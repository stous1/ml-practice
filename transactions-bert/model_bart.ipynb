```python
from transformers import pipeline
```

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
```

```python
categorias = ["Supermercado", "Transporte", "Hogar", "Servicios Públicos", "Entretenimiento"]
```

```python
descripcion = "Pago de gas domiciliario en EPM"
```

```python
resultado = classifier(descripcion, candidate_labels=categorias)
print(resultado)
```
